```python
# Circular Variance with time
plt.figure(figsize=(6,4))

Nsites = occupancies_t.shape[2]

discrete_wave = np.array([
    np.exp((2 * np.pi / occupancies_t.shape[2]) * i * 1j)
    for i in range(occupancies_t.shape[2])
])

product = discrete_wave * occupancies_t[:, 0, :]
m1 = np.sum(product, axis=1) / np.sum(occupancies_t[:, 0, :], axis=1)
cvs = -2 * np.log(np.abs(m1))

plt.plot(tau_t[1:], cvs[1:])
plt.xlabel(r'$\tau$')
plt.ylabel('Circular variance')
plt.title('Circular Variance (measure of cluster width)')
plt.tight_layout()
plt.show()
```
